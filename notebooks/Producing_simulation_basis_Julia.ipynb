# Producing basis for Quack reconstruction

```julia
using HDF5
using LaTeXStrings
using Plots
using Format

using Random, Distributions
using EllipsisNotation

using LinearInterpolations
```

```julia
include("../src/QuackSim.jl")
using .QuackSim
# if installed:
#using QuackSim
```

```julia
# photo-electron kinetic energy axis (in eV) and angular axis (in rad)
minW = 95.0
maxW = 155.0
nbinsW = 160
W_axis = collect(LinRange(minW, maxW, nbinsW))
θ_axis = collect(range(0, 2π, step=2π/16.0))[1:16]
```

```julia
# whether to use GPUs
use_gpu = true

# output H5 file
output = "../example/basis.h5"

# ellipticity
epsilon = 1.0   

# gas
gas = "Ne1s"

# lser wavelength in nm
laser_wavelength = 1030


# maximum ponderomotive potentil, in eV
maxUp = 0.4

# linear FEL polarization
polarization=:horizontal    

# beta: Set to 0 for circular polarization, or 2 for linear polarization
beta = 2.0

# how much to oversample the FEL energy axis
oversampling = 1

# number of bins in time and in the ponderomotive potential axis
nbins_time = 128
nbins_Up = 64

# FEL central energy in eV
fel_energy = 1000.0
```

```julia
simulate_amplitude(
                   fel_energy=fel_energy,
                   W_axis=W_axis,
                   theta_axis=θ_axis,
                   laser_wavelength=laser_wavelength,
                   maxUp=maxUp,
                   output_filename=output,
                   gas=gas,
                   nbins_time=nbins_time,
                   nbins_Up=nbins_Up,
                   oversampling=oversampling,
                   use_gpu=use_gpu,
                   beta=beta,
                   polarization=polarization,
                   ellipticity=epsilon,
                   ) 
```
